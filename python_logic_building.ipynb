Examples:

Input: 1234 → Output: 4321

Input: 500 → Output: 5

Input: -123 → Output: -321

```python
def reverse(num):

    original_num = num

    if num <0:
        num = abs(num)
    rev =  0
    while num > 0:
        last_digit = num % 10
        rev = rev*10 +last_digit
        num = num //10

    if original_num < 0:
        rev = rev*-1

    return rev
num = int(input("enter a num"))
reverse(num)
```

0 → 0

Constraints:

Use arithmetic operations and a while loop.
Do not convert the number to a string.
Handle negative numbers correctly.

```python
def sum_of_digits(number):

    sum = 0
    
    if number <0:
        number = abs(number)
    elif number == 0:
        return sum 

    while number > 0:
        last_digit = number% 10 
        sum = sum + last_digit
        number = number // 10 

    return sum
         

number = int(input("enter a number"))
sum_of_digits(number)
        
        
        
```

[1200, 4500, 3200, 4500, 2800, 1200, 3900]

Write a function that returns the second-largest unique value.

Expected output: 3900

- Requirements
  
Do not use sort() or sorted().

Do not use set().

Handle duplicate values correctly.

Return None if there is no second-largest unique value.

```python

def second_largest(values):
    largest = None
    second_largest = None

    for num in values:
        if largest is None or num > largest:
            second_largest = largest
            largest = num

        elif num != largest and (
            second_largest is None or num > second_largest
        ):
            second_largest = num

    return second_largest


# Test cases
print(second_largest([1200, 4500, 3200, 4500, 2800, 1200, 3900]))  # 3900
print(second_largest([5, 5, 5]))                                    # None
print(second_largest([10, 20]))                                     # 10
print(second_largest([]))                                           # None
print(second_largest([-5, -10, -2]))                                # -5
print(second_largest([5, 3, 5]))                                    # 3
print(second_largest([5, 3, 4]))                                    # 4
```
